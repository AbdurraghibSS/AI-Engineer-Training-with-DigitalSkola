## Belajar Fungsi & OOP pada Python

### 1. Function (`def`)

```python
# Fungsi tanpa parameter & tanpa return -> otomatis mengembalikan None
def salam_pembuka():
    print("Halo, selamat belajar Python!")

hasil = salam_pembuka()
print("Isi hasil:", hasil)
print("Tipe hasil:", type(hasil))
```

```python
# Fungsi dengan parameter dan return value
def hitung_harga_akhir(harga, diskon):
    potongan = harga * diskon
    return harga - potongan

print("Harga akhir:", hitung_harga_akhir(178000, 0.15))
```

```python
# Default parameter & keyword argument
def sapa(nama, waktu="pagi"):
    return f"Selamat {waktu}, {nama}!"

print(sapa("Andi"))                      # pakai nilai default
print(sapa("Budi", "malam"))             # positional argument
print(sapa(waktu="siang", nama="Caca"))  # keyword argument, urutan bebas
```

```python
# Return lebih dari satu nilai -> dikemas sebagai tuple
def statistik_nilai(daftar_nilai):
    rata_rata = sum(daftar_nilai) / len(daftar_nilai)
    return min(daftar_nilai), max(daftar_nilai), rata_rata

terendah, tertinggi, rata = statistik_nilai([80, 75, 90, 65])
print(f"Terendah: {terendah}, Tertinggi: {tertinggi}, Rata-rata: {rata}")
```

```python
# *args (banyak argumen posisi) & **kwargs (banyak argumen bernama)
def total_belanja(*harga):
    return sum(harga)

def buat_profil(**data):
    for kunci, nilai in data.items():
        print(f"{kunci}: {nilai}")

print("Total belanja:", total_belanja(15000, 23000, 7500))
buat_profil(nama="Dewi", kota="Bandung", umur=24)
```

```python
# Lambda: fungsi singkat satu baris, sering dipakai sebagai key saat sorting
peserta = [("Andi", 80), ("Budi", 92), ("Caca", 75)]
urut_nilai = sorted(peserta, key=lambda x: x[1], reverse=True)
print(urut_nilai)
```

### 2. Class & Instance

```python
# Class = cetakan (blueprint), instance = objek nyata hasil cetakan
class Rekening:
    bank = "Bank Nusantara"   # class attribute: sama untuk semua instance

    def __init__(self, pemilik, saldo=0):
        # instance attribute: milik masing-masing objek
        self.pemilik = pemilik
        self.saldo = saldo

    def setor(self, jumlah):
        self.saldo += jumlah
        return f"{self.pemilik} setor {jumlah}. Saldo: {self.saldo}"

    def tarik(self, jumlah):
        if jumlah > self.saldo:
            return f"Gagal: saldo {self.pemilik} tidak cukup."
        self.saldo -= jumlah
        return f"{self.pemilik} tarik {jumlah}. Saldo: {self.saldo}"

    def __str__(self):
        # dipanggil saat objek di-print
        return f"Rekening({self.pemilik}, saldo={self.saldo})"


rek_andi = Rekening("Andi", 100000)
rek_budi = Rekening("Budi")

print(rek_andi.setor(50000))
print(rek_andi.tarik(500000))
print(rek_budi.setor(20000))

# Dua instance punya data sendiri-sendiri, tapi berbagi class attribute
print(rek_andi, "|", rek_budi)
print("Bank:", rek_andi.bank, "&", rek_budi.bank)
```

### 3. Inheritance & Method

#### Inheritance (pewarisan) & override

```python
# Parent class
class Notifikasi:
    def __init__(self, penerima):
        self.penerima = penerima

    def kirim(self, pesan):
        return f"[UMUM] ke {self.penerima}: {pesan}"


# Child class mewarisi Notifikasi, lalu menimpa (override) method kirim
class NotifikasiEmail(Notifikasi):
    def __init__(self, penerima, subjek):
        super().__init__(penerima)   # panggil __init__ milik parent
        self.subjek = subjek

    def kirim(self, pesan):
        return f"[EMAIL] ke {self.penerima} | Subjek: {self.subjek} | {pesan}"


class NotifikasiWhatsApp(Notifikasi):
    def kirim(self, pesan):
        return f"[WA] ke {self.penerima}: {pesan} 📱"


# Polymorphism: method yang sama (kirim), hasil berbeda sesuai class-nya
daftar_notif = [
    Notifikasi("Andi"),
    NotifikasiEmail("budi@mail.com", "Info Kelas"),
    NotifikasiWhatsApp("0812xxxx"),
]
for notif in daftar_notif:
    print(notif.kirim("Kelas dimulai pukul 19.15 WIB"))

print(isinstance(daftar_notif[1], Notifikasi))   # True, karena child dari Notifikasi
```

#### Instance method, class method, static method

```python
class Suhu:
    jumlah_data = 0   # class attribute untuk menghitung objek yang dibuat

    def __init__(self, celsius):
        self.celsius = celsius
        Suhu.jumlah_data += 1

    # Instance method: pakai self, bekerja dengan data milik objek
    def ke_fahrenheit(self):
        return self.celsius * 9 / 5 + 32

    # Class method: pakai cls, sering dipakai sebagai "constructor alternatif"
    @classmethod
    def dari_fahrenheit(cls, fahrenheit):
        return cls((fahrenheit - 32) * 5 / 9)

    # Static method: tidak pakai self/cls, fungsi biasa yang "dititipkan" di class
    @staticmethod
    def wajar(celsius):
        return -10 <= celsius <= 45


jakarta = Suhu(32)
london = Suhu.dari_fahrenheit(50)

print(f"Jakarta: {jakarta.celsius}°C = {jakarta.ke_fahrenheit()}°F")
print(f"London : {london.celsius:.1f}°C")
print("Jumlah data suhu:", Suhu.jumlah_data)
print("Apakah 60°C wajar?", Suhu.wajar(60))
```

### 4. Type Hinting

```python
# Type hint = anotasi tipe data, membantu keterbacaan & autocomplete di IDE
nama_kota: str = "Surabaya"
curah_hujan: list[float] = [12.5, 0.0, 30.2]
suhu_kota: dict[str, float] = {"Jakarta": 32.1, "Bandung": 24.3}


def rata_rata(angka: list[float]) -> float:
    return sum(angka) / len(angka)


def cari_suhu(kota: str, data: dict[str, float]) -> float | None:
    # float | None artinya: bisa float, bisa juga None (kalau kota tidak ada)
    return data.get(kota)


print("Rata-rata curah hujan:", rata_rata(curah_hujan))
print("Suhu Bandung:", cari_suhu("Bandung", suhu_kota))
print("Suhu Medan:", cari_suhu("Medan", suhu_kota))
```

```python
# Type hint TIDAK dicek saat program berjalan (runtime)
def kali_dua(x: int) -> int:
    return x * 2

print(kali_dua(5))        # 10, sesuai harapan
print(kali_dua("ha"))     # 'haha', tetap jalan walau bukan int
# -> Untuk validasi tipe yang benar-benar dicek saat runtime, pakai Pydantic
```

### 5. Pydantic

Install dulu jika belum ada: `pip install pydantic`

```python
from pydantic import BaseModel, Field, ValidationError


class PrakiraanCuaca(BaseModel):
    kota: str
    suhu: float
    kelembapan: int = Field(ge=0, le=100, description="Persentase 0-100")
    hujan: bool = False   # nilai default


# Data valid: "31.5" (string) otomatis dikonversi menjadi float (coercion)
cuaca = PrakiraanCuaca(kota="Semarang", suhu="31.5", kelembapan=78)
print(cuaca)
print("Tipe suhu:", type(cuaca.suhu))
```

```python
# Data tidak valid -> ValidationError, lengkap dengan field yang salah
try:
    PrakiraanCuaca(kota="Bogor", suhu="panas", kelembapan=130)
except ValidationError as e:
    print(e)
```

```python
# Konversi model <-> dict / JSON
print(cuaca.model_dump())        # ke dict
print(cuaca.model_dump_json())   # ke string JSON

teks_json = '{"kota": "Makassar", "suhu": 29, "kelembapan": 85, "hujan": true}'
cuaca2 = PrakiraanCuaca.model_validate_json(teks_json)   # dari string JSON
print(cuaca2)
```

### 6. Mini Studi Kasus: Validasi Output LLM dengan Pydantic

Skenario: LLM diminta membaca berita cuaca lalu mengembalikan JSON. Pydantic memastikan struktur & nilainya sesuai sebelum dipakai di aplikasi.

```python
from typing import Literal


class Wilayah(BaseModel):
    nama: str
    provinsi: str


class PeringatanCuaca(BaseModel):
    jenis: Literal["hujan lebat", "angin kencang", "gelombang tinggi"]
    level: Literal["waspada", "siaga", "awas"]
    wilayah: list[Wilayah]
    keyakinan: float = Field(ge=0.0, le=1.0)
    ringkasan: str | None = None


# Anggap ini jawaban mentah dari LLM (berbentuk string JSON)
jawaban_llm = """
{
  "jenis": "hujan lebat",
  "level": "siaga",
  "wilayah": [
    {"nama": "Kota Bogor", "provinsi": "Jawa Barat"},
    {"nama": "Kota Depok", "provinsi": "Jawa Barat"}
  ],
  "keyakinan": 0.87,
  "ringkasan": "Hujan lebat disertai petir diprakirakan sore hingga malam."
}
"""

peringatan = PeringatanCuaca.model_validate_json(jawaban_llm)
print(f"Peringatan {peringatan.jenis.upper()} - level {peringatan.level}")
for i, w in enumerate(peringatan.wilayah, 1):
    print(f"  {i}. {w.nama}, {w.provinsi}")
print("Keyakinan:", peringatan.keyakinan)
```

```python
# Kalau LLM "ngarang" nilai di luar pilihan, langsung ketahuan
jawaban_salah = '{"jenis": "badai salju", "level": "bahaya", "wilayah": [], "keyakinan": 1.4}'

try:
    PeringatanCuaca.model_validate_json(jawaban_salah)
except ValidationError as e:
    print(f"Ditemukan {e.error_count()} kesalahan:")
    for err in e.errors():
        print(" -", err["loc"], "->", err["msg"])
```
