### 3. Collections

#### List

```python
# inisiasi list
nilai = [80, 80, 75, 50]

# appending single element
nilai.append(85)

# extending list with multiple elements
nilai.extend([90, 95])

print(nilai)
```

#### Tuple

```python
# Inisialisasi & Akses Tuple
tp = ("Number", 122, False)
print("Indeks 0:", tp[0])
print("Indeks 1:", tp[1])
print("Indeks 2:", tp[2])

# Konversi List ke Tuple
list_to_tuple = tuple([1, 2, 3])
print("Tuple hasil konversi:", list_to_tuple)
```

#### Set

```python
# Eliminasi Duplikat
set1     = {11, 222, 222, 33, 33}
print("Set unik:", set1) # Output: {11, 222, 33}
    
# Update dengan List
set1.update([4, 5])
print("Set setelah update list:", set1)

# Update dengan String (String diurai menjadi karakter unik)
set2_as_string = {1, 2}
set2_as_string.update('Yo, What is going on ? ')
print("Set setelah update string:", set2_as_string)
```

#### Dictionary

```python
nilai_mahasiswa = {"Andi": 80, "Budi": 75, "Citra": 90}
print(nilai_mahasiswa)
print("Nilai Andi:", nilai_mahasiswa["Andi"])

nilai_mahasiswa["Dewi"] = 85
nilai_mahasiswa.update({"Budi": 70})
print(nilai_mahasiswa)
```

```python
# Konversi Pasangan Tuple ke Dict
nilai_mahasiswa = dict([
    ('id', 1),
    ('name', 'Andi'),
    ('score', 80),
    ('grade', 'A')
])
print("Dict dari tuple:", nilai_mahasiswa)
```

### 4. Control Flow and Loops

```python
# Control flow, conditional statements

# Data disimpan sebagai dictionary
data = {
    "admin": {"is_admin": True},
    "user": {"is_admin": False}
}

username = input("Enter your username: ")

# Cek apakah username ada di dalam data
if username in data:
    # Cek nilai is_admin-nya
    if data[username]["is_admin"]:
        print("Welcome, admin! (is_admin = True)")
    else:
        print("Welcome, user! (is_admin = False)")
else:
    print("Access denied: Username tidak ditemukan.")
```

```python
# For Loop dengan enumerate
items = ['Andi', 'Budi', 'Caca']
print("\n--- Iterasi For Loop ---")
for idx, x in enumerate(items):
    print(f"Index {idx}: {x}")
```

```python
# While Loop
print("\n--- Iterasi While Loop ---")
count = 0
while count < 3:
    print(f"Count: {count}")
    count += 1
```
